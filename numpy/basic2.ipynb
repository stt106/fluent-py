
# 5. Boolean indexing and Boolean methods
*Chapter 4.1: Boolean indexing; Chapter 4.4: Boolean methods*

## Mental model

A Boolean mask answers a question at each position.

**Filtering** uses those answers to select data. **Reduction** summarizes those answers.

## Summary

- Combine array conditions using `&`, `|`, and `~`.
- Parenthesize individual comparisons.
- Reading through Boolean indexing produces a copy.
- Direct indexed assignment, such as `a[mask] = 0`, changes `a`.
- For Boolean arrays, `sum` counts true values.
- `any` asks whether at least one condition is true; `all` asks whether every condition is true. Both can operate along an axis. ([numpy.org](https://numpy.org/doc/stable/user/basics.indexing.html))

# 6. Fancy indexing
*Chapter 4.1: integer-array indexing*

## Mental model

Integer-array indexing is a request for **specific positions**, potentially reordered or repeated.

## Summary

- `a[[2, 0]]` selects positions in the requested order.
- Reading with advanced integer indexing returns a copy.
- Two matching one-dimensional index arrays select **pairs of coordinates**, not a rectangular block.
- `np.ix_` constructs indices for selecting all combinations of specified rows and columns.
- Direct advanced-index assignment still modifies the target array. ([numpy.org](https://numpy.org/doc/stable/user/basics.indexing.html))

# 7. Transposing, swapping axes, and reshaping
*Chapter 4.1: transposing arrays and swapping axes*

## Mental model

**Transpose changes the order of axes. Reshape changes how the element sequence is grouped.**

Those are different operations.

## Summary

- For a two-dimensional array, `.T` exchanges rows and columns.
- For higher-dimensional arrays, `.T` reverses the axis order.
- `transpose(...)` specifies an axis permutation.
- `swapaxes(...)` exchanges two axes.
- Transposing a one-dimensional array leaves its shape unchanged.
- Reshaping may return a view or require a copy; do not assume either universally. ([numpy.org](https://numpy.org/doc/stable/reference/generated/numpy.transpose.html))

# 15. Linear algebra
*Chapter 4.6*

## Mental model

Elementwise multiplication combines corresponding entries.

Matrix multiplication combines entries through **sums of products**.

## Summary

For compatible two-dimensional arrays:

```text
(m, n) @ (n, k) → (m, k)
```

For a matrix and one-dimensional vector:

```text
(m, n) @ (n,) → (m,)
```

The linear-algebra tools worth recognizing include:

- `solve`: square linear systems
- `lstsq`: least-squares problems
- `inv` and `pinv`: inverse and pseudoinverse
- `qr`, `svd`: decompositions
- `eig`, `eigh`: eigenvalue problems
- `det`: determinant

For this first pass, prioritize understanding multiplication and solving systems over memorizing decomposition APIs. ([numpy.org](https://numpy.org/doc/stable/reference/routines.linalg.html))

# 16. Random walks and simulation
*Chapter 4.7: one walk and many walks*

## Mental model

Separate:

1. **Increments:** changes at each step.
2. **Positions:** cumulative changes from the starting position.
3. **Events:** conditions satisfied by positions.
4. **Summaries:** frequencies and first-event times.

## Summary

A random walk combines generation, cumulative sums, Boolean conditions, and reductions.

For arrays shaped `(simulation, step)`, cumulative summation along `axis=1` accumulates within each simulation.

A crucial edge case: `argmax` returns the first maximum. On an all-false Boolean row, that is index zero—not an indication that the event occurred. Pair first-hit calculations with an existence check using `any`. ([numpy.org](https://numpy.org/doc/stable/reference/generated/numpy.cumsum.html))
