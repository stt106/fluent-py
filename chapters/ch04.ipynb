```python
# Byte Essentials

# Two binary sequences: bytes (immutable) and bytearray (mutable)
# Each item in binary sequence is an int from 0 to 255.
# A slice of a binary sequence always produces a slice of the same binary sequence of the same type (even a slice of 1)
cafe = bytes(s, encoding='utf-8')
cafe
```

```python
# A single item bytes is an int; a slice of bytes is also bytes; b'c' means it's bytes (raw binary) 
# rather than a regular string.
cafe[0], cafe[:1], cafe[:1][0] # Note that a slice of bytes with length 1 is also bytes.


#NOTE: Apart from str where s[0] == s[:1], all other types of sequences they're not the same!
#NOTE: and for built-in sequence types, a slice of a sequence is the same type as the sequence.
```

```python
cafe_arr = bytearray(cafe)
cafe_arr[-1:]
```

```python
# One way to build a binary sequence is calling the ctor with a buffer-like object (e.g. implements buffer protocol).
# This always copies the bytes from the source object to the newly created binary sequence; memoryview doesn't copy!
# This is a low level operation and may involve type casting.
import array
numbers = array.array('h', [-2, -1, 0, 1, 2])
octets = bytes(numbers) # 10 bytes representing the 5 short integers
```
