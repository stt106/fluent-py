## Q13: `bisect` for Sorted Sequences

You have a sorted list of response-time thresholds (in ms) and corresponding performance labels:

```python
thresholds = [50, 200, 500, 1000]
labels = ['excellent', 'good', 'acceptable', 'slow', 'critical']
```

Write a function `classify_latency(ms)` that returns the correct label using `bisect`. For example, `classify_latency(150)` → `'good'`, `classify_latency(1000)` → `'critical'`.

````python
# Your answer here
import bisect
thresholds = [50, 200, 500, 1000]
labels = ['excellent', 'good', 'acceptable', 'slow', 'critical']
def classify_latency(ms):
    i = bisect.bisect(thresholds, ms)
    return labels[i]

for ms in [150, 1000]:
    print(classify_latency(ms))
````

## Q15: Cartesian Product with Listcomps

Using a list comprehension (not `itertools.product`), produce a list of all `(topping, crust)` combinations where:

```python
toppings = ['mushroom', 'pepperoni', 'olive']
crusts = ['thin', 'thick']
```

The result should be ordered with crust as the outer loop: all toppings on thin first, then all toppings on thick.

````python
# Your answer here
toppings = ['mushroom', 'pepperoni', 'olive']
crusts = ['thin', 'thick']

comb = [(crust, topping) for crust in crusts for topping in toppings]
comb
                
````
