```python
import collections

Card = collections.namedtuple('Card', ['rank', 'suit'])

class FrenchDeck:
    suits = ('spades', 'diamonds', 'hearts', 'clubs')
    ranks = tuple(str(i) for i in range(2, 11)) + tuple('JQKA')

    def __init__(self):
        self._cards = tuple(Card(r, s) for s in self.suits for r in self.ranks)

    def __len__(self):
        return len(self._cards)

    def __getitem__(self, key):
        return self._cards[key]


french_deck = FrenchDeck()
```

```python
french_deck[-1]
```

```python
for card in french_deck[9::13]:
    print(card)
```

```python
from random import choice
for _ in range(3):
    print(choice(french_deck))
```

```python
suit_values = dict(spades=3, hearts=2, diamonds=1, clubs=0)
def high_spades(card):
    rank_value = FrenchDeck.ranks.index(card.rank)
    return rank_value*len(suit_values) + suit_values[card.suit]

for card in sorted(french_deck, key=high_spades)[:12]:
    print(card)
```

```python
from math import hypot

class Vector:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def __abs__(self):
        return hypot(self.x, self.y)
    
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __bool__(self):
        return bool(self.x or self.y)

    def __mul__(self, other):
        return Vector(self.x * other, self.y*other)

    def __repr__(self):
        return f'Vector({self.x!r}, {self.y!r})'

v1 = Vector(2, 5)
v2 = Vector(3, 4)
```

```python
v1*2
```

```python
abs(v1), abs(v2)
```
