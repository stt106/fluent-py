```python
RAW = [
    "1700000000,STN-A,21.4,ok",
    "1700000060,STN-B,34.9,ok",
    "1700000120,STN-A,-5.2,ok",
    "1700000180,STN-C,41.1,warn",
    "1700000240,STN-B,18.0,ok",
    "1700000300,STN-A,55.3,fail",
    "1700000360,STN-C,9.7,ok",
    "1700000420,STN-B,27.5,ok",
]

# Task 1 
from collections import namedtuple
Reading = namedtuple('Reading', ['ts', 'station', 'celsius', 'status'])
readings = [Reading(int(ts), station, float(celsius), status) # NOTE: convert the type during unpacking!
            for ts, station, celsius, status in (line.split(',') for line in RAW)]

# Task 2
from array import array
temp_list = [r.celsius for r in readings]
temp_array = array('d', temp_list)
import sys
print(f'Temperature list size: {sum(sys.getsizeof(f) for f in temp_list)}')
print(f'Temperature array size: {sys.getsizeof(temp_array)}')
print('Temperature array size is much smaller because it stores the float values direclty in memory whereas list temperature stores float object references')


# Task 3 
from collections import deque
live_feed = deque(maxlen=3)
for reading in readings:
    live_feed.append(reading.celsius)
    print(list(live_feed))



# Task 4.
BREAKPOINTS = [0, 15, 30, 45]           # ascending
LEVELS = ['freezing', 'cold', 'normal', 'warm', 'critical']
def classify(temp):
    index = bisect.bisect(BREAKPOINTS, temp)
    return LEVELS[index]

for temp in [-5.2, 21.4, 55.3]:
    print(classify(temp))
```

```python
# Task 5 — Top-5 sorted leaderboard.
# Uses readings from Task 1. As you stream through the readings, maintain a list holding the 5 highest celsius values seen so far, kept in ascending sorted order at all times — without ever calling sorted(). Print the final list.
# Correct output: the 5 highest temperatures, ascending.
sorted_temp = []
for reading in readings:
    bisect.insort(sorted_temp, reading.celsius)
    if len(sorted_temp) > 5:
        sorted_temp.pop(0)
print(sorted_temp)


# Task 6 — Query dispatcher with match/case.
# Uses readings from Task 1. Write handle(cmd) where cmd is a list, matching these shapes:

# ["max", station] → the Reading for that station with the highest celsius.
# ["range", lo, hi] → all readings with lo <= celsius <= hi, inclusive.
# ["stats"] → the count of readings and their mean celsius.
# anything else → "unknown command".
# Call it with ["max", "STN-A"], ["range", 20, 40], ["stats"], ["bogus"].
def handle(cmd):
    match cmd:
        case ["max", station]:
            matches = [r for r in readings if r.station == station]
            print(max(matches, key=lambda r: r.celsius) if matches else None)
            # max_temp = float('-inf')
            # result_reading = None
            # for r in readings:
            #     if r.station == station and r.celsius > max_temp:
            #         max_temp = r.celsius
            #         result_reading = r
            # print(result_reading)
        case ["range", lo, hi]:
            result = [r for r in readings if lo <= r.celsius <= hi]
            print(result)
        case ["stats"]:
            count = len(readings)
            total_celsius = sum(r.celsius for r in readings)
            print(count, total_celsius/count)
        case _:
            print("unknown command")

handle(['max', 'STN-A'])
handle(['range', 20, 40])
handle(['stats'])
handle(['bogus'])

# Task 7 — Non-mutating rank.
# Uses readings from Task 1. Produce a new list of all readings ordered hottest→coldest by celsius. Print it, then print readings and confirm the original order is unchanged.
sorted_readings = sorted(readings, key=lambda r: r.celsius, reverse=True)
print(sorted_readings)
print(readings)
```
