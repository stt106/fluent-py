```python
# __missing__ is called when __getitem__ can't find the key. It's NOT defined in the dict base class but any subclass
# implementing it, the standard __getitem__ will call it whenever a key is not found, instead of raising a KeyError.

# A search like `k in my_dict.keys()` is efficient even for large maps 
# because keys() returns a view which is like a set. However, `k in my_dict` does the same job faster because it avoids
# attribute lookup to find .keys() method.
```

```python
# ChainMap
from collections import ChainMap
d1 = dict(a=1, b=2, c=3)
d2 = dict(b=4, d=5)
chain = ChainMap(d1, d2)
```

```python
chain
```

```python
chain['b'] # returns the first instance
```

```python
chain['b'] = 10 # updates or insertion affects the first instance
chain['e'] = 20 # only d1 is inserted!
d1, d2 # d2 is unchanged
```

```python
chain2 = ChainMap(d2, d1)
chain2['b'] # now it returns b from d2 which is the first in the chain!
```

```python
# To create a custom map type, it's better to subclass collections.UserDict rather than built-in dict which 
# has some implementation shortcut that forces us to override methods that we can just inherit from UserDict
# without problem.
# NOTE that UserDict doesn't inherit from dict, it uses a instance field called data as a dict internally e.g. composition!
import collections
class StrKeyDict(collections.UserDict):
    def __missing__(self, key):
        if isinstance(key, str):
            raise KeyError(key)
        return self.data[str[key]]

    def __contains__(self, key):
        return str(key) in self.data

    def __setitem__(self, key, item):
        self.data[str(key)] = item
```

```python
# Dictionary Views
# dict.keys(), dict.items(), dict.values() are all dictionary views that are read-only; those three types are internal
# so no instance of them can be created directly. 
d = dict(a=10, b=20, c=30)
values = d.values()
list(values)
```

```python
d['d'] = 40
values # it's a proxy to dict values so it sees the update
```

```python
# dict_values can be reversed but not indexing, whereas dict_keys and dict_items implement more set like methods.
# It's almost like FrozenSet!
for v in reversed(values):
    print(v)
```

```python
values[0]
```

```python
# In CPython, instances of the same class can share the internal structure containing their attribute names. 
# Each instance then stores mainly its own values. This is called a key-sharing dictionary optimization.
# It is specifically about the attributes stored in an object’s __dict__. 
# It is a memory optimization guideline, not a Python language requirement. 
# NOTE: The most practical rule is: initialize the normal set of instance attributes in __init__, 
# NOTE: and add attributes later only when the design genuinely calls for dynamic object shape.

# NOTE DO THIS
class User:
    def __init__(self, name):
        self.name = name
        self.email = None       # establish this attribute immediately

    def set_email(self, email):
        self.email = email      # update an existing attribute
```

```python
# NOTE: DON'T DO THIS!
class User:
    def __init__(self, name):
        self.name = name

    def set_email(self, email):
        self.email = email      # creates a new attribute later and this reduces the effectiveness of optimization and
        # use more memory across many instances.
```
