```python
# Random walk in OO

import random

class RandomWalk:
    def __init__(self):
        self.position = 0

    def walk(self, n):
        self.position = 0
        for _ in range(n):
            yield self.position
            self.position += 2*random.randint(0, 1) - 1

from timeit import timeit
walker = RandomWalk()
%timeit [position for position in walker.walk(n=10_000)]

        
```

```python
def random_walk(n):
    position = 0
    walk = [position]
    for _ in range(n):
        position += 2*random.randint(0, 1) - 1
        walk.append(position)
    return walk

%timeit random_walk(10_000)
```

```python
from itertools import accumulate

def random_walk_faster(n):
    steps = random.choices([-1, 1], k=n)
    return [0]+list(accumulate(steps))

%timeit random_walk_faster(10_000)
```

```python
import numpy as np

def random_walk_fastest(n=10_000):
    steps = np.random.choice([-1, 1], size=n)
    return np.cumsum(steps)

%timeit random_walk_fastest()
```
